# **Fundamentos e Tipos de Dados **

Estatística Descritiva vs. Inferencial: A Descritiva tira uma "foto" do que já aconteceu (ex.: calcular a média de vendas do mês passado). A Inferencial usa uma parte dos dados para tentar "prever" o todo ou o futuro (ex.: pesquisa eleitoral).

População vs. Amostra: População é todo mundo (todos os 5.000 clientes da loja). Amostra é um pedaço representativo (200 clientes que responderam à pesquisa).


Tipos de Variáveis (Como o Python enxerga as colunas):

- Qualitativa (Categórica): É texto ou categoria (ex.: cor do produto, nível de satisfação: bom/médio/ruim).

- Quantitativa Discreta: É número inteiro que a gente conta (ex.: número de filhos, quantidade de visitas no site).

- Quantitativa Contínua: É número que a gente mede e aceita vírgula (ex.: altura, peso, temperatura, valor da venda em R$).

# **Formatos de Distribuição e Correlação**

- Distribuição Normal (Curva de Sino): A maioria fica no meio, e poucos nas pontas.
Aqui, Média = Mediana = Moda (ex.: altura das pessoas, notas de uma prova equilibrada).

- Distribuição Assimétrica (Skewed): Tem uma "cauda" esticada para um lado por causa de valores extremos.

 Assimétrica à Direita (Positiva): Muitos valores baixos e poucos muito altos puxando a média para cima (Média > Mediana). Ex.: renda salarial.

Assimétrica à Esquerda (Negativa): Muitos valores altos e poucos muito baixos puxando a média para baixo (Média < Mediana). Ex.: idade de aposentadoria.

Distribuição Bimodal: Tem dois "morros" (picos). Indica que existem dois grupos diferentes misturados nos dados (ex.: tempo de maratona com atletas profissionais e amadores).

- Correlação (r): Mede se duas variáveis andam juntas (vai de -1 a +1). Positiva: tráfego no site sobe e vendas sobem.
Negativa: preço sobe e vendas caem. Lembrete importante: correlação não significa que uma coisa causou a outra!

A Média (mean)

Imagine que você e 4 amigos ganharam quantidades diferentes de balas. Um ganhou 2, outro 8, outro 10... A média é o ato de juntar todas as balas em um único monte e dividir exatamente por igual entre todo mundo.
